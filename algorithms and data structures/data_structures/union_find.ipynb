# Quick Find
- O(1) for `find()` : Store root instead of parent
- O(N) for `union()`: Needs to adjust roots by traversal instead of backtracking parents

```python
class UnionFind:
    def __init__(self, size):
        self.root = [i for i in range(size)]

    def find(self, x):
        return self.root[x] # O(1)

    def union(self, x, y):
        rootX = self.find(x)
        rootY = self.find(y)

        if rootX != rootY:
            # Make roots same
            for i in range(len(self.root)):
                if self.root[i] == rootY:
                    self.root[i] = rootX
    
    def connected(self, x, y):
        return self.find(x) == self.find(y)

# Test Case
uf = UnionFind(10)
# 1-2-5-6-7 3-8-9 4
uf.union(1, 2)
uf.union(2, 5)
uf.union(5, 6)
uf.union(6, 7)
uf.union(3, 8)
uf.union(8, 9)
print(uf.connected(1, 5))  # true
print(uf.connected(5, 7))  # true
print(uf.connected(4, 9))  # false
# 1-2-5-6-7 3-8-9-4
uf.union(9, 4)
print(uf.connected(4, 9))  # true
print(uf.root)
```

# Quick Union
- O(N) for `find()` : Traverse every vertex to find root for input vertex
- O(N) for `union()`: Connect to root instead of parent. `find()` roots for both nodes and updating the array value at index 

```python
class UnionFind:
    def __init__(self, size):
        self.root = [i for i in range(size)]
    
    def find(self, x):
        while(x != self.root[x]):
            x = self.root[x] # Search its parent next
        return x
    
    def union(self, x, y):
        rootX = self.find(x)
        rootY = self.find(y)

        if rootX != rootY:
            self.root[rootY] = rootX
    
    def connected(self, x, y):
        return self.find(x) == self.find(y)



# Test Case
uf = UnionFind(7)
# {0,1,2,3} and {4,5,6}
uf.union(0, 1)
uf.union(1, 2)
uf.union(1, 3)
uf.union(4, 5)
uf.union(4, 6)
print(uf.connected(0, 3))  # true
print(uf.connected(4, 6))  # true
print(uf.connected(1, 5)) # false
uf.union(1, 5)
# {0,1,2,3,4,5,6}
print(uf.connected(1, 5)) # true
print(uf.root)
```

# Union By Rank
- **PROBLEM** : In Quick Union, repeated unions can create tall skewed trees (like a linked list). In the worst case, find() becomes O(N) because we must traverse the entire chain to reach the root.
- **SOLUTION** : Use Union by Rank. Always attach the tree with smaller rank under the tree with larger rank to keep the structure balanced.
- `union find constructor` : O(N)
- `find(m)` : O(logN) - Worst case if both the ranks being unioned are same
- `union(x,y)` : O(logN) - Defined by find()

# Path Compression Optimization
- Flatten the graph in the `find()` function reducing the complexity overtime to near constant

```python
class UnionFind:
    def __init__(self, size):
        self.root = [i for i in range(size)]
        self.rank = [1] * size

    def find(self, x):
        while x != self.root[x]:
            x = self.root[x]
        return x
    
    def union(self, x, y):
        rootX = self.find(x)
        rootY = self.find(y)

        if rootX != rootY:
            if self.rank[rootX] > self.rank[rootY]:
                # X is larger than Y, absorb Y
                self.root[rootY] = rootX
            elif self.rank[rootX] < self.rank[rootY]:
                self.root[rootX] = rootY
            else:
                self.root[rootY] = rootX
                self.rank[rootX] += 1
    
    def connected(self, x, y):
        return self.find(x) == self.find(y)

# Test Case
uf = UnionFind(10)
# 1-2-5-6-7 3-8-9 4
uf.union(1, 2)
uf.union(2, 5)
uf.union(5, 6)
uf.union(6, 7)
uf.union(3, 8)
uf.union(8, 9)
print(uf.connected(1, 5))  # true
print(uf.connected(5, 7))  # true
print(uf.connected(4, 9))  # false
# 1-2-5-6-7 3-8-9-4
uf.union(9, 4)
print(uf.connected(4, 9))  # true
```
