# Convert to Base 7
Given an integer num, return a string of its base 7 representation.

Input: num = 100
Output: "202"

```python
def convertToBase7(num):
    s = ""
    is_negative = False
    if(num<0) : is_negative = True

    if(num != 0):
        if(is_negative):
            num = num * -1

        while(num!=0):
            remainder = num // 7
            quotient = num % 7
            s += str(quotient)
            num = remainder

        if is_negative:
            return "-" + s[::-1]
        return s[::-1]
    else:
        return "0"

print(convertToBase7(100))
```

# Convert to Hexadecimal
Given an integer num, return a string of its hexadecimal representation.

```python
def toHex(num):
    if num == 0:
        return "0"
    
    digits = "0123456789abcdef"
    result = ""
    negative = num < 0

    num = abs(num)
    print(num)

    while num > 0:
        remainder = num % 16
        result = digits[remainder] + result
        print(result)
        num = num // 16

    return "-" + result if negative else result

print(toHex(-1))
```
