## Longest Common Subsequence
https://leetcode.com/problems/longest-common-subsequence/

**Framework**:
1. State variable - dp[i][j] longest subsequence we can get using first i chars from text1 and first j chars from text2.
2. Recurrence relation -
   - dp[i][j] = dp[i-1][j-1] + 1 if the last characters matched.
   - dp[i][j] = max(dp[i-1][j], dp[i][j-1]) if last characters didnt match. It means we need to get the point where they last matched. This works because subsequence `doesnt need to be contiguous`.
3. Base case - 
   Cant have common substrings if either is empty
   - dp[0][j] = 0 for all j 
   - dp[i][0] = 0 for all i


```python
def longestCommonSubsequence(text1, text2):
    n = len(text1)
    m = len(text2)

    # NOTE : Watch out for indexes everywhere
    dp = [[0] * (m + 1) for _ in range(n + 1)]

    # Recurrence
    for i in range(1, n+1):
        for j in range(1, m+1):
            if text1[i-1] == text2[j-1]:
                dp[i][j] = dp[i-1][j-1] + 1
            else:
                dp[i][j] = max(dp[i-1][j], dp[i][j-1])
    
    return dp[n][m]

text1 = "ace"
text2 = "bafcdfde"
print(longestCommonSubsequence(text1, text2))
```

## Longest Common Substring

**Framework**:
1. State variable - dp[i][j] longest substring we can get using first i chars from text1 and first j chars from text2.
2. Recurrence relation -
   - dp[i][j] = dp[i-1][j-1] + 1 if the last characters matched.
   - dp[i][j] = 0, if last previous didnt match, as substring demands order and cant have breaks unlike subsequences.
   - Max length = `max(Max length, dp[i][j])`
3. Base case - 
   Cant have common substrings if either is empty
   - dp[0][j] = 0 for all j 
   - dp[i][0] = 0 for all i
   - Max length = 0

```python
def longestCommonSubstring(text1, text2):
    n = len(text1)
    m = len(text2)

    # NOTE : Watch out for indexes everywhere
    dp = [[0] * (m + 1) for _ in range(n + 1)]
    max_len = 0

    # Recurrence
    for i in range(1, n+1):
        for j in range(1, m+1):
            if text1[i-1] == text2[j-1]:
                dp[i][j] = dp[i-1][j-1] + 1
                max_len = max(max_len, dp[i][j])
    
    return max_len

text1 = "aceface" # face and ac
text2 = "bafacedfdeac" # face and ac
print(longestCommonSubstring(text1, text2))
```
